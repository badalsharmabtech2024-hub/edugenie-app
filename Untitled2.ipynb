```python
import pandas as pd

# Replace 'your_file_name.csv' with the actual path to your CSV file
csv_file_path = 'air_quality_historical.csv'

try:
    df = pd.read_csv(csv_file_path)
    print(f"Successfully loaded '{csv_file_path}' into a DataFrame.")
    print("First 5 rows:")
    display(df.head())
except FileNotFoundError:
    print(f"Error: The file '{csv_file_path}' was not found. Please check the file path.")
except Exception as e:
    print(f"An error occurred while reading the CSV file: {e}")
```

Let's calculate the mean, median, and mode for a specific column. You can change `column_name` to any column of interest.

```python
# Calculate Mean, Median, and Mode for the 'pm10' column
column_name = 'pm10'

if column_name in df.columns:
    # Calculate Mean
    mean_value = df[column_name].mean()
    print(f"Mean of '{column_name}': {mean_value:.2f}")

    # Calculate Median
    median_value = df[column_name].median()
    print(f"Median of '{column_name}': {median_value:.2f}")

    # Calculate Mode (a column can have multiple modes)
    mode_value = df[column_name].mode()
    print(f"Mode(s) of '{column_name}':")
    display(mode_value)
else:
    print(f"Error: Column '{column_name}' not found in the DataFrame.")
```
